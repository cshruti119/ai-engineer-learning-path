```python
import os
from dotenv import load_dotenv
from anthropic import Anthropic
load_dotenv()
client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
model = "claude-sonnet-4-6"
```

```python
question1 = "What is the best way to plan a trip?"


response1     = client.messages.create(
    model=model,
    max_tokens=300,
    messages=[{"role": "user", "content": question1}],
)

print("Question:", question1)
print("Response:", response1.content[0].text)
```

```python
question2 = "Write another line"
response2 = client.messages.create(
    model=model,
    max_tokens=300,
    messages=[{"role": "user", "content": question2}],
)

print("Question:", question2)
print("Response:", response2.content[0].text)
```

```python
def add_user_message(messages, content):
    userMessage = {"role": "user", "content": content}
    messages.append(userMessage)

def add_assistant_message(messages, content):
    assistantMessage = {"role": "assistant", "content": content}
    messages.append(assistantMessage)

def chat(messages):
    response = client.messages.create(
        model=model,
        max_tokens=300,
        messages=messages,
    )
    return response.content[0].text
    
```

```python
messages = []
add_user_message(messages, "Define load balancing in simple terms and in one sentence.")
response = chat(messages)
add_assistant_message(messages, response)
add_user_message(messages, "Add one more line.")
response = chat(messages)
response
```

Chat bot exercise 

```python
chat_messages = []

while True:
    user_input = input(">")
    if(user_input.lower() == "exit"):
        break
    print(">:", user_input)
    print("--------------------------------")
    add_user_message(chat_messages, user_input)
    response = chat(chat_messages)
    print("Response:", response)
    print("--------------------------------")
    add_assistant_message(chat_messages, response)

```
